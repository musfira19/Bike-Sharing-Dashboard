## Menentukan Pertanyaan Bisnis

**SMART Question** adalah sebuah framework untuk merumuskan pertanyaan secara terstruktur agar memeperoleh informasi yang mendalam.

* **Spesific (Spesifik)**
  * Pertanyaan harus jelas, fokus pada sebuah topik tertentu, dan tidak bermakna ganda. Hindari pertanyaan yang terlalu luas.
    * Salah: Bagaimana cara meningkatkan penjualan?
    * Benar: Faktor apa saja yang memengaruhi penurunan penjualan produk kategori elektronik di wilayah Jakarta selama kuartal terakhir?
* **Measurable (Terukur)**
  * Pertanyaan harus bisa dijawab dengan angka atau matrix yang konkret, harus tahu apa yang akan dihitung.
    * Salah: Apakah pelanggan senang dengan layanan kita?
    * Benar: Berapa skor rata-rata Customer Satisfaction untuk layanan purna jual bulan ini dibandingkan bulan lalu?
* **Action-Oriented (Berorientasi Aksi)**
  * Hasil dari pertanyaan harus bisa memberikan arahan untuk melakukan tindakan nyata. Jika pertanyaan terjawab, stakeholder harus tahu apa langkah selanjutnya.
    * Salah: Mengapa orang suka berbelanja?
    * Benar: Fitur apa pada aplikasi yang paling sering digunakan sebelum pengguna memutuskan untuk melakukan checkout?
* **Relevant (Relevan)**
  * Hasil dari pertanyaan harus sejalan dengan tujuan utama bisnis atau masalah yang sedang dihadapi.
    * Salah: Menanyakan tentang stok gudang saat masalah utamanya adalah efektivitas kampanye media sosial.
    * Benar: Apakah kampanye iklan di Instagram memberikan Return on Ad Spend (ROAS) yang lebih tinggi dibandingkan iklan di TikTok?
* **Time-bound (Terikat Waktu)**
  * Pertanyaan harus ada batasan waktu yang jelas agar analisis memiliki konteks yang tepat.
    * Salah: Berapa banyak pengguna baru kita?
    * Benar: Berapa tingkat pertumbuhan pengguna baru secara bulanan (Month-over-Month) sepanjang tahun 2025?

**Contoh pertanyaan bisnis yang memenuhi seluruh elemen SMART**

***"Faktor apa saja yang memengaruhi penurunan conversion rate pada pengguna aplikasi Android di wilayah Jabodetabek sebesar 5% selama periode Flash Sale Maret 2026?"***

Keterangan:

- **Specific**: Fokus pada "penurunan conversion rate" untuk "aplikasi Android" di "Jabodetabek". Bukan sekadar penjualan turun.
- **Measurable**: Ada angka konkret yang ingin dianalisis, yaitu penurunan sebesar "5%".
- **Action-Oriented**: Dengan mengetahui faktor penyebabnya misalnya bug pada versi Android tertentu atau kendala logistik di Jabodetabek, tim bisa langsung melakukan perbaikan teknis atau operasional.
- **Relevant**: Penurunan konversi saat Flash Sale adalah masalah kritis bagi bisnis retail/e-commerce.
- **Time-bound**: Dibatasi pada periode spesifik "Flash Sale Maret 2026".

- **Pertanyaan 1:**  Bagaimana perbedaan rata-rata jumlah penyewaan sepeda antara hari kerja dan hari nonkerja selama periode 2011–2012?
- **Pertanyaan 2:** Bagaimana perbedaan rata-rata jumlah penyewaan sepeda berdasarkan musim dan kondisi cuaca selama periode 2011–2012?
- **Pertanyaan 3:** Pada jam berapa jumlah penyewaan sepeda mencapai puncak dan titik terendah selama periode pengamatan tahun 2011–2012, dan bagaimana pola tersebut berbeda antara hari kerja dan hari nonkerja?

### Cleaning Data

Pada tahap ini, dilakukan pembersihan data berdasarkan permasalahan yang ditemukan pada tahap Assessing Data.
```python
# Kolom kategorikal pada masing-masing dataset
categorical_cols_day = [
    "season", "yr", "mnth", "holiday",
    "weekday", "workingday", "weathersit"]

categorical_cols_hour = [
    "season", "yr", "mnth", "hr", "holiday",
    "weekday", "workingday", "weathersit"]

# Mengubah tipe data menjadi category
day_df[categorical_cols_day] = day_df[categorical_cols_day].astype("category")
hour_df[categorical_cols_hour] = hour_df[categorical_cols_hour].astype("category")

day_df.info()
hour_df.info()
```

## Visualization & Explanatory Analysis

### Pertanyaan 1:
Bagaimana perbedaan rata-rata jumlah penyewaan sepeda antara hari kerja dan hari nonkerja selama 2011–2012?

```python
# Mengelompokkan jumlah penyewaan berdasarkan hari kerja
workingday_analysis = day_df.groupby("workingday", observed=True)["cnt"].agg(
    ["mean", "median", "min", "max", "sum"]).reset_index()

workingday_analysis["workingday"] = workingday_analysis["workingday"].map({
    0: "Hari Nonkerja",
    1: "Hari Kerja"})

display(workingday_analysis)
```

```python
# Visualisasi
plt.figure(figsize=(8, 5))

sns.barplot(
    data=day_df,
    x="workingday",
    y="cnt",
    estimator="mean",
    errorbar=None)

plt.title("Rata-rata Penyewaan Sepeda Berdasarkan Jenis Hari")
plt.xlabel("Jenis Hari")
plt.ylabel("Rata-rata Jumlah Penyewaan")
plt.xticks([0, 1], ["Hari Nonkerja", "Hari Kerja"])
plt.tight_layout()
plt.show()
```

**Insight:** Berdasarkan hasil pengelompokan data, rata-rata jumlah penyewaan sepeda pada hari kerja mencapai 4.584,82 penyewaan per hari, sedangkan pada hari nonkerja sebesar 4.330,17 penyewaan per hari. Hal ini menunjukkan bahwa rata-rata penyewaan sepeda pada hari kerja lebih tinggi sekitar 254,65 penyewaan dibandingkan hari nonkerja.

Nilai median juga menunjukkan pola yang serupa, yaitu 4.582 pada hari kerja dan 4.459 pada hari nonkerja. Sementara itu, total penyewaan pada hari kerja mencapai 2.292.410, lebih tinggi dibandingkan hari nonkerja sebesar 1.000.269. Perbedaan total tersebut perlu dilihat dengan mempertimbangkan jumlah hari pada masing-masing kategori. Secara deskriptif, hasil ini menunjukkan bahwa aktivitas penyewaan sepeda cenderung lebih tinggi pada hari kerja dibandingkan hari nonkerja selama periode pengamatan 2011–2012.

### Pertanyaan 2:
Bagaimana perbedaan rata-rata jumlah penyewaan sepeda berdasarkan musim dan kondisi cuaca selama 2011–2012?

```python
# Rata-rata penyewaan berdasarkan musim
season_analysis = day_df.groupby("season", observed=True)["cnt"].agg(
    ["mean", "median", "min", "max", "sum"]
).reset_index()

season_analysis["season"] = season_analysis["season"].map({
    1: "Spring",
    2: "Summer",
    3: "Fall",
    4: "Winter"})

display(season_analysis)
```

```python
# Rata-rata penyewaan berdasarkan kondisi cuaca
weather_analysis = day_df.groupby("weathersit", observed=True)["cnt"].agg(
    ["mean", "median", "min", "max", "sum"]
).reset_index()

weather_analysis["weathersit"] = weather_analysis["weathersit"].map({
    1: "Clear / Few Clouds",
    2: "Mist / Cloudy",
    3: "Light Snow / Rain",
    4: "Heavy Rain / Snow / Fog"})

display(weather_analysis)
```

```python
# Visualisasi
plt.figure(figsize=(8, 5))

sns.barplot(
    data=day_df,
    x="season",
    y="cnt",
    estimator="mean",
    errorbar=None
)

plt.title("Rata-rata Penyewaan Sepeda Berdasarkan Musim")
plt.xlabel("Musim")
plt.ylabel("Rata-rata Jumlah Penyewaan")
plt.xticks(
    [0, 1, 2, 3],
    ["Musim Semi", "Musim Panas", "Musim Gugur", "Musim Dingin"]
)

plt.tight_layout()
plt.show()
```

**Insight:**
- Berdasarkan hasil pengelompokan data berdasarkan musim, rata-rata jumlah penyewaan sepeda tertinggi terjadi pada musim Fall sebesar 5.644,30 penyewaan per hari, sedangkan rata-rata terendah terjadi pada musim Spring sebesar 2.604,13 penyewaan per hari. Musim Summer memiliki rata-rata 4.992,33 penyewaan per hari, sementara Winter sebesar 4.728,16 penyewaan per hari. Hasil tersebut menunjukkan adanya perbedaan rata-rata penyewaan sepeda antar musim. Penyewaan cenderung lebih tinggi pada musim Fall dan lebih rendah pada musim Spring. Namun, hasil ini merupakan gambaran deskriptif dan belum menunjukkan bahwa musim menjadi penyebab langsung perubahan jumlah penyewaan.
- Berdasarkan kondisi cuaca, rata-rata jumlah penyewaan sepeda tertinggi terjadi saat cuaca Clear / Few Clouds, yaitu sebesar 4.876,79 penyewaan per hari. Sementara itu, kondisi Mist / Cloudy memiliki rata-rata 4.035,86 penyewaan per hari dan kondisi Light Snow / Rain memiliki rata-rata terendah sebesar 1.803,29 penyewaan per hari. Hasil ini menunjukkan bahwa rata-rata penyewaan sepeda cenderung lebih tinggi pada kondisi cuaca cerah dibandingkan kondisi berkabut atau berawan serta kondisi dengan salju ringan atau hujan. Meskipun demikian, hasil ini bersifat deskriptif sehingga belum dapat digunakan untuk menyimpulkan hubungan sebab-akibat.



```python
# Mengelompokkan penyewaan berdasarkan jam dan jenis hari
hour_analysis = hour_df.groupby(
    ["workingday", "hr"], observed=True
)["cnt"].mean().reset_index()

hour_analysis["workingday"] = hour_analysis["workingday"].map({
    0: "Hari Nonkerja",
    1: "Hari Kerja"
})

display(hour_analysis)
```

```python
# Jam dengan rata-rata penyewaan tertinggi dan terendah
peak_hour = hour_analysis.loc[
    hour_analysis.groupby("workingday", observed=True)["cnt"].idxmax()]

lowest_hour = hour_analysis.loc[
    hour_analysis.groupby("workingday", observed=True)["cnt"].idxmin()]

print("Jam dengan rata-rata penyewaan tertinggi:")
display(peak_hour)

print("Jam dengan rata-rata penyewaan terendah:")
display(lowest_hour)
```

```python
# Visualisasi rata-rata penyewaan sepeda berdasarkan kondisi cuaca
plt.figure(figsize=(8, 5))

sns.barplot(
    data=day_df,
    x="weathersit",
    y="cnt",
    estimator="mean",
    errorbar=None)

plt.title("Rata-rata Penyewaan Sepeda Berdasarkan Kondisi Cuaca")
plt.xlabel("Kondisi Cuaca")
plt.ylabel("Rata-rata Jumlah Penyewaan")
plt.xticks(
    [0, 1, 2],
    ["Cerah", "Berkabut/Berawan", "Hujan Ringan"])

plt.tight_layout()
plt.show()
```

**Insight:** Berdasarkan hasil analisis rata-rata jumlah penyewaan sepeda berdasarkan jam, terdapat perbedaan waktu puncak penyewaan antara hari kerja dan hari nonkerja. Pada hari kerja, rata-rata penyewaan tertinggi terjadi pada pukul 17.00 sebesar 525,29 penyewaan, sedangkan rata-rata terendah terjadi pada pukul 03.00 sebesar 4,94 penyewaan.

Sementara itu, pada hari nonkerja, rata-rata penyewaan tertinggi terjadi pada pukul 13.00 sebesar 372,73 penyewaan dan rata-rata terendah terjadi pada pukul 04.00 sebesar 8,26 penyewaan. Pola ini menunjukkan bahwa penyewaan pada hari kerja cenderung mencapai puncak pada sore hari, sedangkan pada hari nonkerja mencapai puncak pada siang hari. Pada kedua kategori hari, penyewaan relatif rendah pada dini hari.

Hasil ini menunjukkan adanya perbedaan pola waktu penyewaan antara hari kerja dan hari nonkerja selama periode pengamatan 2011–2012.

### Pertanyaan 3:
Pada jam berapa jumlah penyewaan sepeda mencapai puncak dan titik terendah selama periode pengamatan tahun 2011–2012, dan bagaimana pola tersebut berbeda antara hari kerja dan hari nonkerja?

```python
# Menghitung rata-rata penyewaan per jam berdasarkan jenis hari
hour_analysis = (
    hour_df.groupby(["hr", "workingday"], observed=True)["cnt"]
    .mean()
    .reset_index()
)

# Memberi label jenis hari agar mudah dipahami
hour_analysis["Jenis Hari"] = hour_analysis["workingday"].map({
    1: "Hari Kerja",
    0: "Hari Nonkerja"
})

# Membuat line chart
plt.figure(figsize=(12, 6))

sns.lineplot(
    data=hour_analysis,
    x="hr",
    y="cnt",
    hue="Jenis Hari",
    hue_order=["Hari Kerja", "Hari Nonkerja"],
    marker="o",
    linewidth=2.5,
    estimator=None
)

plt.title("Pola Rata-rata Penyewaan Sepeda Berdasarkan Jam")
plt.xlabel("Jam dalam Sehari")
plt.ylabel("Rata-rata Jumlah Penyewaan")
plt.xticks(range(24))
plt.grid(axis="y", linestyle="--", alpha=0.3)
plt.legend(title="Jenis Hari")
plt.tight_layout()
plt.show()

```

**Insight:**
- Hari Kerja (Garis Biru): Memiliki pola dua puncak (bimodal) yang sangat tajam, yang menunjukkan kaitan erat dengan jam berangkat dan pulang kerja/sekolah.
- Hari Nonkerja (Garis Oranye): Memiliki pola satu puncak (unimodal) yang lebih landai dan terdistribusi merata di siang hari.
- Puncak Pagi: Lonjakan penyewaan tertinggi terjadi pada jam 08.00 pagi (mendekati 500 penyewaan), yang mengindikasikan penggunaan sepeda untuk mobilitas pagi hari menuju tempat kerja atau institusi pendidikan.
- Puncak Sore: Lonjakan terbesar sepanjang hari terjadi pada jam 17.00 (5 sore) dengan angka di atas 500 penyewaan, menyusul penurunan bertahap hingga malam hari.
- Penyewaan mulai meningkat secara konsisten sejak pukul 08.00 pagi dan mencapai puncak tertinggi antara jam 12.00 hingga 14.00 (siang hari) yang berkisar di angka 370-an penyewaan.
- Pola ini mencerminkan penggunaan sepeda untuk keperluan rekreasi, santai, atau aktivitas luar ruangan di siang hari.
- Pada kisaran pukul 10.00 pagi dan pukul 16.00 sore, jumlah penyewaan pada hari nonkerja mulai melampaui hari kerja sebelum akhirnya melonjak tajam kembali di jam pulang kerja
- Baik pada hari kerja maupun nonkerja, tingkat penyewaan sepeda berada di titik terendah pada dini hari, yaitu antara pukul 02.00 hingga pukul 05.00 pagi.

## Analisis Lanjutan (Opsional)

```python
# Memilih variabel numerik yang akan dianalisis
correlation_cols = ["temp", "atemp", "hum", "windspeed", "cnt"]

# Menghitung korelasi Pearson
correlation_matrix = day_df[correlation_cols].corr(method="pearson")

# Menampilkan matriks korelasi
correlation_matrix
```

```python
plt.figure(figsize=(10, 7))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5,
    vmin=-1,
    vmax=1,
    center=0
)

plt.title("Correlation Heatmap Kondisi Lingkungan dan Penyewaan Sepeda")
plt.tight_layout()
plt.show()
```

```python
# Mengurutkan korelasi terhadap jumlah penyewaan
cnt_correlation = correlation_matrix["cnt"].drop("cnt").sort_values(
    ascending=False)

print("Korelasi Variabel terhadap Jumlah Penyewaan:")
print(cnt_correlation)
```

**Insight:**
- atemp (0,51) Memiliki korelasi positif terkuat terhadap jumlah penyewaan sepeda (cnt), yang menunjukkan bahwa kenyamanan suhu yang dirasakan pengguna adalah faktor lingkungan paling dominan dalam meningkatkan minat sewa.
- temp (0,28) Menunjukkan korelasi positif yang moderat terhadap penyewaan, yang berarti suhu aktual tetap berdampak baik, meski pengaruhnya tidak sebesar suhu yang dirasakan secara subjektif (atemp).
- hum (-0,04) Memiliki nilai korelasi yang sangat mendekati nol, mengindikasikan bahwa tingkat kelembapan udara tidak memiliki hubungan linier yang signifikan terhadap naik turunnya penyewaan sepeda.
- windspeed (-0,01) Memiliki korelasi yang praktis nol, menandakan bahwa kecepatan angin tidak menjadi faktor penentu yang mempengaruhi volume penyewaan sepeda harian.
